```python
def mahalanobis(x, z, M):
  transposed = np.transpose(x - z) 
  scaler = np.linalg.matmul(M, (x - z))
  tmp = np.linalg.matmul(transposed, scaler)
  return np.sqrt(tmp.sum())
```

```python
from scipy.spatial import distance
import numpy as np

s = np.array([3, 3, 3])

S = np.diag(s)

M = S @ S # s^2 for mahalanobis since (x_hat - z_hat)^2 (I guess)

x = np.array([1, 0, 0])
z = np.array([0, 1, 0])

mahalanobis = mahalanobis(x, z, M)
euclidean = distance.euclidean(x * s, z * s)

print(mahalanobis)
print(euclidean)
```
