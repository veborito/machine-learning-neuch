```python
import numpy as np
```

```python
# We have n models 
# Each model gives us a probability for rain, or some event i
# This can be a table: likelihood[event, model]
# For every model, we have a prior belief: prior[model]
# The posterior calculation is simple:
# prior * likelihood / marginal

def get_posterior(prior, likelihood, event):
    n_model = len(prior)
    product = [prior[model]*likelihood[event, model] for model in range(n_models)]
    marginal = sum(product)
    return product / marginal
```

```python
n_events = 2
n_models = 3
likelihood = np.zeros([n_events, n_models])
```

```python
likelihood[1, 0] = 0.2
likelihood[1, 1] = 0.3
likelihood[1, 2] = 0.5
likelihood[0] = 1 - likelihood[1]
likelihood
```

```python
prior = np.array([0.5, 0.4999999, 0.0000001])
print(prior)

posterior = get_posterior(prior, likelihood, 1)
print("posterior: ", posterior)
print("likelihood: ", likelihood[1])
prior = posterior
    
```
